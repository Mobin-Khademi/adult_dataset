```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth,association_rules
```

```python
# read data
df = pd.read_csv("../dist/adult_preprocessed.csv").astype(bool)
```

```python
# fpgrowth
frequent_items = fpgrowth(df,(13000/len(df)),use_colnames=True)
# association rules
rule = association_rules(frequent_items,metric="confidence",min_threshold=0.95)
```

```python
#write frequent items on txt
with open("../dist/frequent_itemsets.txt", "w", encoding='utf-8') as f:
    for _, row in frequent_items.sort_values('support', ascending=False).iterrows():
        items = ' , '.join(list(row['itemsets']))
        f.write(items + "\n")

#write rules on txt
with open("../dist/rules.txt","w") as f:
    for _ , row in rule.sort_values('lift',ascending=False).iterrows():
        a = tuple(row['antecedents'])
        b = tuple(row['consequents'])
        f.write(str(a) + ' -> ' + str(b) + '\n')
```

```python
print("frequent_items : ",len(frequent_items))
print("rules : ",len(rule))
```
